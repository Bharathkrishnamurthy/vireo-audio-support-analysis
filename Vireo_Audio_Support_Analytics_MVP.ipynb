````python
agent_metrics = (
    tickets.groupby("agent_id")
    .agg(
        tickets=("ticket_id", "count"),
        csat_responses=("csat_score", "count"),
        avg_csat=("csat_score", "mean"),
        median_aht_min=("handle_time_min", "median"),
        avg_aht_min=("handle_time_min", "mean")
    )
    .reset_index()
)

agent_metrics["low_csat_rate"] = (
    tickets.groupby("agent_id")["csat_score"]
    .apply(lambda x: (x.dropna() <= 2).mean())
    .values
)

display(agent_metrics.head(10))
````

````python
agent_context = (
    agents.groupby("agent_id")
    .agg(
        teams=("team", lambda x: " | ".join(x.dropna().astype(str).unique())),
        tiers=("tier", lambda x: " | ".join(x.dropna().astype(str).unique())),
        sites=("site", lambda x: " | ".join(x.dropna().astype(str).unique())),
        shifts=("shift", lambda x: " | ".join(x.dropna().astype(str).unique()))
    )
    .reset_index()
)

agent_metrics_context = agent_metrics.merge(
    agent_context,
    on="agent_id",
    how="left"
)

display(agent_metrics_context.head(10))
````

````python
# Step 23 — Business baseline

total_tickets = len(tickets)
csat_responses = tickets["csat_score"].notna().sum()
avg_csat = tickets["csat_score"].mean()
low_csat_count = (tickets["csat_score"] <= 2).sum()
low_csat_rate = (tickets["csat_score"] <= 2).mean()

weekly_tickets = 650
monthly_tickets = weekly_tickets * 52 / 12

print(f"Total tickets in dataset: {total_tickets:,}")
print(f"CSAT responses: {csat_responses:,}")
print(f"Average CSAT: {avg_csat:.2f}/5")
print(f"Low CSAT responses (1-2): {low_csat_count:,}")
print(f"Low CSAT rate: {low_csat_rate:.1%}")
print(f"Estimated monthly tickets: {monthly_tickets:,.0f}")
````

````python
# Step 24 — Validated CSAT baseline

csat_responses = tickets["csat_score"].notna().sum()
low_csat_count = (tickets["csat_score"] <= 2).sum()

csat_response_rate = csat_responses / len(tickets)
low_csat_rate_rated = low_csat_count / csat_responses

print(f"Total tickets: {len(tickets):,}")
print(f"CSAT responses: {csat_responses:,}")
print(f"CSAT response rate: {csat_response_rate:.1%}")
print(f"Average CSAT: {tickets['csat_score'].mean():.2f}/5")
print(f"Low CSAT responses: {low_csat_count:,}")
print(f"Low CSAT rate among rated tickets: {low_csat_rate_rated:.1%}")
print(f"Monthly ticket volume: {monthly_tickets:,.0f}")
````

````python
# Step 25 — First-response SLA performance

tickets["first_response_at"] = pd.to_datetime(
    tickets["first_response_at"],
    errors="coerce"
)
tickets["created_at"] = pd.to_datetime(
    tickets["created_at"],
    errors="coerce"
)

tickets["first_response_min"] = (
    tickets["first_response_at"] - tickets["created_at"]
).dt.total_seconds() / 60

sla_minutes = {
    "chat": 15,
    "voice": 120,
    "social": 240,
    "email": 480
}

tickets["sla_target_min"] = tickets["channel"].map(sla_minutes)

tickets["sla_breach"] = (
    tickets["first_response_min"] > tickets["sla_target_min"]
)

sla_summary = (
    tickets
    .groupby("channel")
    .agg(
        tickets=("ticket_id", "count"),
        breaches=("sla_breach", "sum")
    )
    .reset_index()
)

sla_summary["breach_rate"] = (
    sla_summary["breaches"] /
    sla_summary["tickets"]
)

display(sla_summary)
````

````python
# Step 26 — SLA credit scenario at current ticket volume

total_breaches = int(tickets["sla_breach"].sum())
historical_credit = total_breaches * 350

historical_months = 18
monthly_breaches = total_breaches / historical_months
monthly_credit = monthly_breaches * 350

annual_credit = monthly_credit * 12

print(f"Historical SLA breaches: {total_breaches:,}")
print(f"Historical SLA credits: ₹{historical_credit:,.0f}")
print(f"Average breaches/month: {monthly_breaches:,.1f}")
print(f"Estimated SLA credits/month: ₹{monthly_credit:,.0f}")
print(f"Estimated SLA credits/year: ₹{annual_credit:,.0f}")
````

````python
# Step 27 — SLA improvement scenarios

scenario_rates = [0.10, 0.20, 0.30, 0.50]

scenario = pd.DataFrame({
    "breach_reduction": scenario_rates
})

scenario["avoided_breaches_month"] = (
    monthly_breaches * scenario["breach_reduction"]
)

scenario["avoided_credit_month"] = (
    scenario["avoided_breaches_month"] * 350
)

scenario["avoided_credit_year"] = (
    scenario["avoided_credit_month"] * 12
)

scenario["remaining_credit_year"] = (
    annual_credit - scenario["avoided_credit_year"]
)

display(scenario)
````

````python
# Step 28 — Handle-time baseline

valid_aht = tickets[
    tickets["handle_time_min"].notna() &
    (tickets["handle_time_min"] >= 0)
].copy()

print(f"Valid AHT records: {len(valid_aht):,}")
print(f"Median AHT: {valid_aht['handle_time_min'].median():,.1f} min")
print(f"Mean AHT: {valid_aht['handle_time_min'].mean():,.1f} min")
print(f"90th percentile: {valid_aht['handle_time_min'].quantile(0.90):,.1f} min")
print(f"95th percentile: {valid_aht['handle_time_min'].quantile(0.95):,.1f} min")
````

````python
# Step 29 — AHT by team

team_aht = (
    valid_aht
    .groupby("assigned_team")
    .agg(
        tickets=("ticket_id", "count"),
        median_aht_min=("handle_time_min", "median"),
        mean_aht_min=("handle_time_min", "mean"),
        p90_aht_min=("handle_time_min", lambda x: x.quantile(0.90))
    )
    .reset_index()
)

display(
    team_aht.sort_values(
        "median_aht_min",
        ascending=False
    )
)
````

````python
# Step 30 — Final review-priority table

final_review = (
    review_priority
    .head(10)
    [
        [
            "agent_id",
            "teams",
            "tiers",
            "tickets",
            "csat_responses",
            "avg_csat",
            "low_csat_rate",
            "median_aht_min",
            "sites",
            "shifts"
        ]
    ]
    .copy()
)

final_review = final_review.rename(columns={
    "teams": "team",
    "tiers": "tier",
    "avg_csat": "avg_csat",
    "low_csat_rate": "low_csat_rate",
    "median_aht_min": "median_aht_min"
})

display(final_review)
````

````python
# Step 31 — Bottom-10 contribution to low CSAT

bottom10_tickets = tickets[
    tickets["agent_id"].isin(bottom10_ids)
].copy()

bottom10_csat = bottom10_tickets[
    bottom10_tickets["csat_score"].notna()
].copy()

bottom10_low = (
    bottom10_csat["csat_score"] <= 2
).sum()

overall_low = (
    tickets["csat_score"] <= 2
).sum()

print(f"Bottom-10 CSAT responses: {len(bottom10_csat):,}")
print(f"Bottom-10 low-CSAT responses: {bottom10_low:,}")
print(f"Overall low-CSAT responses: {overall_low:,}")
print(
    f"Bottom-10 share of all low-CSAT responses: "
    f"{bottom10_low / overall_low:.1%}"
)
````

````python
# Step 32 — CSAT improvement scenario

bottom10_low_rate = bottom10_low / len(bottom10_csat)

for reduction in [0.10, 0.20, 0.30]:
    avoided_low = bottom10_low * reduction
    new_low_total = overall_low - avoided_low

    print(f"{int(reduction*100)}% relative reduction:")
    print(f"  Avoided low-CSAT responses: {avoided_low:.1f}")
    print(f"  Remaining low-CSAT responses: {new_low_total:.1f}")
````

````python
# Step 33 — SLA performance: Bottom-10 vs everyone else

bottom10_mask = tickets["agent_id"].isin(bottom10_ids)

sla_compare = tickets.groupby(bottom10_mask).agg(
    tickets=("ticket_id", "count"),
    sla_breaches=("sla_breach", "sum")
).reset_index()

sla_compare["breach_rate"] = (
    sla_compare["sla_breaches"] / sla_compare["tickets"]
)

sla_compare["sla_credit_exposure_inr"] = (
    sla_compare["sla_breaches"] * 350
)

sla_compare
````

````python
# Step 34 — Bottom-10 category mix

bottom10_category = (
    tickets[bottom10_mask]
    .groupby("category")
    .agg(
        tickets=("ticket_id", "count"),
        csat_responses=("csat_score", "count"),
        avg_csat=("csat_score", "mean")
    )
    .reset_index()
)

bottom10_category["ticket_share"] = (
    bottom10_category["tickets"] / bottom10_category["tickets"].sum()
)

bottom10_category.sort_values("tickets", ascending=False)
````

````python
# Step 35 — Bottom-10 vs overall CSAT by category

overall_category = (
    tickets[tickets["csat_score"].notna()]
    .groupby("category")
    .agg(
        overall_responses=("csat_score", "count"),
        overall_csat=("csat_score", "mean")
    )
    .reset_index()
)

category_compare = bottom10_category.merge(
    overall_category,
    on="category",
    how="left"
)

category_compare["csat_gap"] = (
    category_compare["avg_csat"] -
    category_compare["overall_csat"]
)

category_compare.sort_values("csat_gap")
````

````python
# Step 36 — Where are low-CSAT responses concentrated?

low_csat_category = (
    tickets[tickets["csat_score"] <= 2]
    .groupby("category")
    .size()
    .reset_index(name="low_csat_responses")
)

low_csat_category["share_of_low_csat"] = (
    low_csat_category["low_csat_responses"] / overall_low
)

low_csat_category.sort_values(
    "low_csat_responses",
    ascending=False
)
````

````python
# Step 37 — Bottom-10 share of low-CSAT by category

problem_categories = [
    "Charging & Battery",
    "Delivery & Shipping",
    "Returns & Refunds",
    "Warranty & Repair"
]

problem_low = (
    tickets[tickets["csat_score"] <= 2]
    .groupby("category")
    .size()
    .reset_index(name="overall_low_csat")
)

bottom10_problem_low = (
    tickets[
        bottom10_mask &
        (tickets["csat_score"] <= 2) &
        (tickets["category"].isin(problem_categories))
    ]
    .groupby("category")
    .size()
    .reset_index(name="bottom10_low_csat")
)

problem_compare = (
    problem_low[
        problem_low["category"].isin(problem_categories)
    ]
    .merge(bottom10_problem_low, on="category", how="left")
    .fillna(0)
)

problem_compare["bottom10_share"] = (
    problem_compare["bottom10_low_csat"] /
    problem_compare["overall_low_csat"]
)

problem_compare.sort_values(
    "overall_low_csat",
    ascending=False
)
````

# Vireo Audio — Humanized Findings & Business Observations

## 1. Overall Customer Experience

The dataset contains **11,750 support tickets** across the 18-month period.

Of these, **5,196 tickets received a CSAT response**, giving a CSAT response rate of **44.2%**.

The overall average CSAT is **3.33/5**.

Among customers who responded to the CSAT survey, **1,121 responses were low-CSAT (score 1–2)**. This represents **21.6% of rated tickets**.

### Key Observation

The overall picture is not that every customer is dissatisfied. However, roughly **1 in 5 CSAT responses is negative**, giving Vireo a meaningful group of customer experiences to investigate.

> **Baseline: 21.6% of rated CSAT responses are low-CSAT.**

Because the CSAT response rate is 44.2%, these results represent customers who responded to the survey rather than every support interaction.

---

## 2. Where Customer Dissatisfaction Is Concentrated

Low-CSAT responses are not evenly distributed across support categories.

| Category | Share of All Low-CSAT Responses |
|---|---:|
| Charging & Battery | **24.17%** |
| Delivery & Shipping | **16.86%** |
| Returns & Refunds | **11.60%** |
| Warranty & Repair | **11.15%** |

Together, these four categories account for approximately **63.8% of all observed low-CSAT responses**.

### Humanized Observation

The training opportunity should not be treated only as an agent-by-agent exercise.

A large portion of observed dissatisfaction is concentrated around a relatively small number of customer problem areas.

For example, retraining an agent without improving the troubleshooting or resolution approach for recurring Charging, Delivery, Returns, or Warranty issues may not fully address the underlying customer experience.

### Key Takeaway

> **The training opportunity is both people-focused and problem-focused.**

---

## 3. Bottom-10 Review Priority Group

The requested Bottom-10 analysis produced a **10-agent Tier-1 Review Priority group** with sufficient CSAT response volume.

Together, these agents handled:

- **3,292 tickets**
- **1,442 CSAT responses**
- **384 low-CSAT responses**

The group represents **34.3% of all observed low-CSAT responses**.

### Humanized Observation

This concentration is sufficient to justify using the group as a review queue.

However, the list should not be interpreted as:

> "The 10 worst employees."

The analysis identifies a **review-priority signal**, not individual causation.

Customer satisfaction can also be affected by:

- Issue category
- Team workflow
- Transfers
- Case complexity
- Customer expectations
- Operational constraints

Therefore, individual tickets and closing notes should be reviewed before assigning individual training.

### Key Takeaway

> **Use the Bottom-10 list as a coaching/review queue rather than a disciplinary ranking.**

---

## 4. What Is Happening Inside the Review-Priority Group?

Within the Bottom-10 Review Priority group, the largest ticket categories are:

- **Returns & Refunds**
- **Delivery & Shipping**

Together, these two categories represent **60.6% of the group's tickets**.

Charging & Battery is also important because the group recorded an observed average CSAT of approximately **2.29/5** for this category.

Warranty & Repair had an observed average CSAT of approximately **1.91/5**, but this result is based on only **33 CSAT responses**, so it should be interpreted cautiously.

### Humanized Observation

The useful question is therefore not simply:

> "Which agents should be trained?"

It is:

> **"Which agents should be reviewed, and which customer problems should their training address?"**

The data supports reviewing the priority group particularly around:

1. Returns & Refunds
2. Delivery & Shipping
3. Charging & Battery
4. Warranty & Repair

---

## 5. Bottom-10 Group vs Overall Category Performance

The Bottom-10 Review Priority group had lower observed CSAT than the overall category benchmark in **8 of the 11 categories**.

Some of the larger observed differences were:

It does **not prove that agent behavior caused the lower CSAT**.

The Warranty & Repair comparison should receive additional caution because the review-priority group has only **33 CSAT responses** in that category.

## 7. Handle Time Requires Context

The dataset contains **11,183 valid handle-time records**.

The policy defines handle time as:

> **First response → resolution**

The resulting distribution is highly right-skewed.

| Handle-Time Metric | Result |
|---|---:|
| Valid records | **11,183** |
| Median | **29 minutes** |
| Mean | **1,646.9 minutes** |
| 90th percentile | **4,496.8 minutes** |
| 95th percentile | **7,354 minutes** |

### Humanized Observation

The mean handle time by itself would be misleading because a relatively small number of very long cases create a large right tail.

Different teams also handle fundamentally different types of work.

For example:

- Escalations & Warranty median AHT: **4,570 minutes**
- Returns Desk median AHT: **2,892 minutes**
- Logistics median AHT: **1,469 minutes**
- Chat Frontline median AHT: **23 minutes**
- Email Frontline median AHT: **22 minutes**
- Billing median AHT: **21 minutes**
- Voice Frontline median AHT: **21 minutes**

### Key Takeaway

> **AHT should be interpreted using median values and team context rather than as a simple individual ranking.**

This is particularly important because the support policy distinguishes Tier-1 and Tier-2 workflows.
